```python
def print_state(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def depth_limited_search(state, goal, depth, path, on_path, step_counter):

    step_counter[0] += 1

    print("--------------------------------")
    print("Step:", step_counter[0])
    print("Depth Limit Remaining:", depth)

    if path:
        print("Move:", path[-1][0])
    else:
        print("Starting State")

    print_state(state)

    if state == goal:
        print(">>> GOAL FOUND <<<")
        return path

    if depth == 0:
        print("Depth limit reached.")
        return None

    state_tuple = tuple(state)
    on_path.add(state_tuple)

    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dr, dc, move in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_state = state.copy()

            new_zero = new_row * 3 + new_col

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            new_state_tuple = tuple(new_state)

            if new_state_tuple not in on_path:

                result = depth_limited_search(
                    new_state,
                    goal,
                    depth - 1,
                    path + [(move, new_state.copy())],
                    on_path,
                    step_counter
                )

                if result is not None:
                    return result

    on_path.remove(state_tuple)

    print("Backtracking from this state.")

    return None


def IDS(initial, goal):

    depth = 0

    while True:

        print("\n")
        print("========================================")
        print("ITERATIVE DEEPENING - DEPTH LIMIT:", depth)
        print("========================================")

        step_counter = [0]

        result = depth_limited_search(
            initial,
            goal,
            depth,
            [],
            set(),
            step_counter
        )

        if result is not None:
            print("\nSolution found at depth:", depth)
            print("States explored in this iteration:",
                  step_counter[0])
            return result

        print("\nNo solution at depth:", depth)
        print("States explored:", step_counter[0])

        depth += 1


initial = [
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
]

goal = [
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
]


print("================================")
print("INITIAL STATE")
print("================================")
print_state(initial)

print("================================")
print("GOAL STATE")
print("================================")
print_state(goal)


solution = IDS(initial, goal)


print("\n")
print("================================")
print("FINAL SOLUTION PATH")
print("================================")

print("\nInitial State:")
print_state(initial)

for move, state in solution:
    print("Move:", move)
    print_state(state)

print("Goal Reached!")
print("Number of moves:", len(solution))
```
