```python
import math
import random

N = 8


def cost(state):
    """Number of attacking pairs. state[col] = row of the queen in that column."""
    return sum(
        1
        for i in range(N)
        for j in range(i + 1, N)
        if state[i] == state[j] or abs(state[i] - state[j]) == j - i
    )


def random_neighbour(state):
    """Move one queen to a different row in its own column."""
    neighbour = state[:]
    col = random.randrange(N)
    neighbour[col] = random.choice([r for r in range(N) if r != state[col]])
    return neighbour


def simulated_annealing(initial, T=100.0, alpha=0.995, T_min=0.001):
    current = initial
    # T never reaches exactly 0 when multiplied by alpha, so stop at T_min
    while T > T_min and cost(current) > 0:
        nxt = random_neighbour(current)
        delta_e = cost(current) - cost(nxt)
        if delta_e > 0 or random.random() < math.exp(delta_e / T):
            current = nxt
        T *= alpha
    return current


def print_board(state):
    """Print the board. state may hold fewer than N queens."""
    for row in range(N):
        print(" ".join("Q" if col < len(state) and state[col] == row else "." for col in range(N)))


def show_placement(solution):
    """Place the solved queens on an empty board, one column at a time."""
    for count in range(1, N + 1):
        col = count - 1
        print(f"\nStep {count}: place queen {count} at row {solution[col]}, column {col}")
        print_board(solution[:count])


def solve(max_restarts=100):
    for _ in range(max_restarts):
        initial = [random.randrange(N) for _ in range(N)]
        result = simulated_annealing(initial)
        if cost(result) == 0:
            return result
    return None


if __name__ == "__main__":
    solution = solve()
    if solution:
        show_placement(solution)
        print(f"\nAll {N} queens placed with no attacks: {solution}")
    else:
        print("No solution found")
```
